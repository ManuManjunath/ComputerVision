```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
img = cv2.imread("data/sammy_face.jpg")
# Not doing colour channel correction since we're only performing edge detection.
plt.imshow(img)
```

```python
edges = cv2.Canny(image = img, threshold1 = 127, threshold2 = 127)
# 127 is the middle value between 0 and 255

plt.imshow(edges)
```

```python
# Trying out other threshold values
edges = cv2.Canny(image = img, threshold1 = 0, threshold2 = 255)
# 127 is the middle value between 0 and 255

plt.imshow(edges)
```

```python
edges = cv2.Canny(image = img, threshold1 = 0, threshold2 = 127)
# 127 is the middle value between 0 and 255

plt.imshow(edges)
```

```python
edges = cv2.Canny(image = img, threshold1 = 127, threshold2 = 255)
# 127 is the middle value between 0 and 255

plt.imshow(edges)
```

```python
# Choosing a threshold - General practice

# Calculate median pixel value
med_val = np.median(img)
print(med_val)

# Set lower threshold to either 0 or 70% of median pixel value (Whichever is greater)
lower = int(max(0, 0.7*med_val))
print(lower)
# Set higher threshold to either 255 or 30% over the median pixel value (Whichever is lower)
upper = int(min(255, 1.3*med_val))
print(upper)

edges = cv2.Canny(image = img, threshold1 = lower, threshold2 = upper)

plt.imshow(edges)
```

```python
# Lets do the same after blurring the image first
blurred_img = cv2.blur(img, ksize=(7, 7))

edges = cv2.Canny(image = blurred_img, threshold1 = lower, threshold2 = upper)

plt.imshow(edges)
```
