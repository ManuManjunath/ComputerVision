```python
img = cv2.imread("data/crossword.jpg", 0)
plt.imshow(img, cmap = "gray")
```

```python
def show_pic(img):
    fig = plt.figure(figsize=(15, 15))
    ax = fig.add_subplot(111)
    ax.imshow(img, cmap = "gray")
```

```python
show_pic(img)
```

```python
ret, thr_img1 = cv2.threshold(img, 150, 255, cv2.THRESH_BINARY)
print(ret)
show_pic(thr_img1)
```

```python
# Adaptive threshold
maxval = 255
blocksize = 9 # Takes odd numbers
const = 6
thr_img2 = cv2.adaptiveThreshold(img, maxval, cv2.ADAPTIVE_THRESH_MEAN_C, cv2.THRESH_BINARY, blocksize, const)
show_pic(thr_img2)
```

```python
# Blending thr_img1 and thr_img2
blended = cv2.addWeighted(src1 = thr_img1, alpha = 0.6, src2 = thr_img2, beta = 0.4, gamma = 0)
show_pic(blended)
```
